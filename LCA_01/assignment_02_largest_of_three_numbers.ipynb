```python
#Assignment 2
a, b, c = map(int, input("Enter three numbers").split())
if a >= b and a >= c:
    print(f"largest of the three integers is {a}")
elif b >= a and b >= c:
    print(f"largest of the three integers is {b}")
elif c >= a and c >= b:
    print(f"largest of the three integers is {c}")
```
