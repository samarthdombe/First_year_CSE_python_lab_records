```python
#Assignment 3
def triangle_checker(a, b, c):
    if a + b > c and b + c > a and a + c > b:
        if b*b == (a*a) + (c*c) or (a*a) == (b*b) + (c*c) or (c*c) == (a*a) + (b*b):
            print("The following sides represents a Right angled triangle")
        else:
            print("The following sides represent a triangle")
    else:
        print("The following sides does not represent a triangle")
n1, n2, n3 = map(int, input("Enter three numbers").split())
triangle_checker(n1, n2, n3)
```

```python
#Assignment 5
pan_no = input("Enter your PAN number")
valid = True
if len(pan_no) != 10:
    valid = False
else:
    for i in range (0,5):
        if pan_no[i].islower():
            valid = False
            break
    for i in range (5,9):
        if pan_no[i].isdigit():
            valid = False
            break
    if pan_no[9].isupper():
        valid = True
if valid == True:
    print("Valid PAN number")
else:
    print("Invalid PAN number")
```
