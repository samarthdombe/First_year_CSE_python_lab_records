```python
#Assignment 5
pan_no = input("Enter your PAN number")
valid = True
if len(pan_no) != 10:
    valid = False
else:
    for i in range (0,5):
        if pan_no[i].islower():
            valid = False
            break
    for i in range (5,9):
        if pan_no[i].isdigit():
            valid = False
            break
    if pan_no[9].isupper():
        valid = True
if valid == True:
    print("Valid PAN number")
else:
    print("Invalid PAN number")
```
